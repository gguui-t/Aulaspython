```python
# importando o pandas (bibliotecas)
import pandas as pd

# importando todos os arquivos
vendas_df = pd.read_csv('Vendas2017.csv', sep=';')
vendas_df

clientes_df = pd.read_csv('Clientes.csv', sep=';')
clientes_df

produtos_df = pd.read_csv('Produtos.csv', sep=';')
produtos_df

promocoes_df = pd.read_csv('Promocoes.csv', sep=';')
promocoes_df

lojas_df = pd.read_csv('Lojas.csv', sep=';')
lojas_df
```

```python
# exibindo os conteudos das bases de dados
vendas_df
produtos_df
```

```python
clientes_df = clientes_df[['ID Cliente','E-mail']]
produtos_df = produtos_df [['ID Produto','Categoria']]
lojas_df = lojas_df [['ID Loja','Nome da Loja']]
```

```python
# Mesclando os Dataframes
vendas_df = vendas_df.merge(produtos_df, on='ID Produto')
vendas_df = vendas_df.merge(lojas_df, on='ID Loja')
vendas_df = vendas_df.merge(clientes_df, on='ID Cliente')

display(vendas_df) # pois esta fazendo de varias bases de dados
```

```python
# quantas vezes o cliente comprou com a empresa

frequencia_cliente = vendas_df ['E-mail'].value_counts()
display(frequencia_cliente)

frequencia_cliente.plot() # plot gera o grafico
```

```python
frequencia_cliente[:5].plot(figsize=(15,5)) # figsize ajusta os heixos do grafico
```
