```python
def cadastrar_produtos(): # sempre nome de funçao tem que esta no imfinitivo 
    produto = input('Digite o nome do produto a ser cadastrado: ')
    produto = produto.casefold() # funcao casefold diminiu as letras
    produto = produto.strip() # retira espaços em branco desnecessario na informaçao
    print(' produto {} cadastrado com sucesso!!' .format(produto))
    return produto
produto = cadastrar_produtos()

print(produto)

```
